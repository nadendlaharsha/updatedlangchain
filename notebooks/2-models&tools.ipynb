# LangChain Model Initialization Examples (Gemini & Groq)

```python
import os
from dotenv import load_dotenv

load_dotenv()
os.environ['GOOGLE_API_KEY'] = os.getenv('GOOGLE_API_KEY')
os.environ['GROQ_API_KEY'] = os.getenv('GROQ_API_KEY')
```

## 1. Google Gemini via init_chat_model

```python
from langchain.chat_models import init_chat_model

gemini_model = init_chat_model("gemini-1.5-flash", model_provider="google_genai")
gemini_model
```

```python
response_gemini = gemini_model.invoke("Hey, how have you been?")
print(response_gemini.content)
```

## 2. Groq via init_chat_model (WORKING)

```python
from langchain.chat_models import init_chat_model

groq_model = init_chat_model("llama-3.3-70b-versatile", model_provider="groq")
groq_model
```

```python
response_groq = groq_model.invoke("Hey, how have you been?")
print(response_groq.content)
```

```python
from langchain_groq import ChatGroq
model=ChatGroq(model='llama-3.3-70b-versatile')
response=model.invoke("why do parrots talk?")
response.text
```

STREAMING

```python
from langchain_groq import ChatGroq
model=ChatGroq(model='llama-3.3-70b-versatile')
for chunk in model.stream("why do parrots talk?"):
    print(chunk.text,end="")
```

BATCH 


```python


model.batch(["what is ai?","what is deep learning?","what is machine learning"],config={"max_concurrency":5,})
```

```python
from langchain.tools import tool

@tool
def get_weather(loc:str)->str:
    """get the weather at location"""
    return f"{loc} is sunny"
model_with_tools=model.bind_tools([get_weather])
```

```python
response=model_with_tools.invoke("whats the weather in boston?")
print(response)
for tool_calls in response.tool_calls:
    print(f"tool:{tool_calls['name']}")
    print(f"args:{tool_calls['args']}")
    
```

tool call loops


```python
messages=[{'role':'user','content':'whats the weather in boston?'}]
ai_msg=model_with_tools.invoke(messages)
messages.append(ai_msg)
for tool_call in ai_msg.tool_calls:
    tool_result=get_weather.invoke(tool_call)
    messages.append(tool_result)
final_response=model_with_tools.invoke(messages)
print(final_response.text)
```
